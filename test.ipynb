```python
import torch
import torch.nn as nn
from torch.nn import functional as F
from dataclasses import dataclass
```

```python
@dataclass
class GPTConfigs:
    blocksize: int = 1024 # Maximum sequence lenght (context window)
    vocabsize: int = 50257 # (Number of possible token)
    n_layer: int = 12 # Number of layers
    n_head: int = 12 # Number of heads
    n_embd:int = 768 # Embedding dimension

config = GPTConfigs()
```

```python
transformer_nul = nn.ModuleDict({
    # 1. Embedding layers (Usually nn.Embedding, but kept as Linear if that is your design)
    "wte": nn.Embedding(config.vocabsize, config.n_embd),
    "wpe": nn.Embedding(config.blocksize, config.n_embd),
    
    # 2. Group the transformer blocks cleanly into a ModuleList
    "h": nn.ModuleList([
        nn.Linear(config.blocksize, config.n_embd) for i in range(config.n_layer)
    ]),
    
    # 3. Final layer norm
    "ln_f": nn.LayerNorm(config.n_embd)
})
transformer_nul["h"][0]
```

```python
transformer = nn.ModuleDict(
	{	
	    "wte": nn.Embedding(config.vocabsize, config.n_embd),
	    "wpe": nn.Embedding(config.blocksize, config.n_embd),
	    **{f"h_{i}": nn.Linear(config.blocksize, config.n_embd) for i in range(config.n_layer)},
	    "ln_f": nn.LayerNorm(config.n_embd, config.n_embd)
	}
)

transformer
```
